# **PHASE 3 — FEATURE SELECTION**

```python
# CHECK COLUMN NAMES
print(df.columns)

# DROP IRRELEVANT COLUMNS
df = df.drop(columns=['ID', 'Z_CostContact', 'Z_Revenue'])

# CREATE ADDITIONAL FEATURES
# Customer Age
df['Age'] = 2025 - df['Year_Birth']

# Total Children
df['Children'] = df['Kidhome'] + df['Teenhome']

# Total Purchases
df['Total_Purchases'] = (
    df['NumWebPurchases'] +
    df['NumCatalogPurchases'] +
    df['NumStorePurchases']
)

# SELECT NUMERICAL COLUMNS
numerical_df = df.select_dtypes(include=['int64', 'float64'])

# CORRELATION MATRIX
correlation_matrix = numerical_df.corr()

# HEATMAP
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(18,12))

sns.heatmap(
    correlation_matrix,
    cmap='coolwarm',
    annot=False
)

plt.title("Correlation Heatmap")

plt.show()

# CORRELATION WITH TOTAL_SPENDING
target_corr = correlation_matrix['Total_Spending'].sort_values(ascending=False)

print("\nCorrelation with Total_Spending:\n")
print(target_corr)

# BAR CHART OF CORRELATIONS
plt.figure(figsize=(12,8))

target_corr.drop('Total_Spending').plot(
    kind='bar'
)

plt.title("Feature Correlation with Total Spending")
plt.xlabel("Features")
plt.ylabel("Correlation")

plt.show()

# SELECT IMPORTANT FEATURES
selected_features = target_corr[
    abs(target_corr) >= 0.10
].index.tolist()

print("\nSelected Features:\n")
print(selected_features)
```

Several irrelevant columns were removed before modelling, including ID, Z_CostContact, and Z_Revenue. The ID column does not contribute to prediction because it only acts as a unique identifier, while the Z_CostContact and Z_Revenue columns contain constant values and therefore provide no meaningful information to the models.

Additional features such as Age, Children, and Total_Purchases were engineered to improve business understanding and better represent customer behavior. These features summarize demographic and purchasing patterns that may influence customer spending and campaign response.

The correlation heatmap shows relationships between numerical variables in the dataset. Variables related to customer income and purchasing activity appear to have stronger positive relationships with Total_Spending, indicating that these features are important predictors for regression modelling.

Based on correlation analysis and business relevance, the most important features were selected for further machine learning tasks. This process helps improve model performance, reduce unnecessary complexity, and enhance interpretability of the results.

# **PHASE 4 — DATA PREPROCESSING**

```python
# CHECKING MISSING VALUES
print("Missing Values:\n")
print(df.isnull().sum())

# HANDLING MISSING VALUES
df['Income'] = df['Income'].fillna(df['Income'].median())

# CHECKING DUPLICATE ROWS
print("\nDuplicate Rows:")
print(df.duplicated().sum())

# DROP DATE COLUMN
df = df.drop(columns=['Dt_Customer'], errors='ignore')

# ENCODE CATEGORICAL VARIABLES

# Education Encoding
education_mapping = {
    'Basic': 0,
    '2n Cycle': 1,
    'Graduation': 2,
    'Master': 3,
    'PhD': 4
}

df['Education'] = df['Education'].replace(education_mapping)

# Marital Status Encoding
df['Marital_Status'] = df['Marital_Status'].replace({
    'Single': 0,
    'Together': 1,
    'Married': 2,
    'Divorced': 3,
    'Widow': 4,
    'Alone': 5,
    'Absurd': 6,
    'YOLO': 7
})

# DEFINE REGRESSION FEATURES & TARGET

X_reg = df.drop(columns=[
    'Total_Spending',
    'Response',
    'MntWines',
    'MntFruits',
    'MntMeatProducts',
    'MntFishProducts',
    'MntSweetProducts',
    'MntGoldProds'
])

y_reg = df['Total_Spending']

# DEFINE CLASSIFICATION FEATURES & TARGET

X_cls = df.drop(columns=['Total_Spending', 'Response'])

y_cls = df['Response']

# TRAIN-TEST SPLIT

from sklearn.model_selection import train_test_split

X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
    X_reg,
    y_reg,
    test_size=0.2,
    random_state=42
)

X_train_cls, X_test_cls, y_train_cls, y_test_cls = train_test_split(
    X_cls,
    y_cls,
    test_size=0.2,
    random_state=42,
    stratify=y_cls
)

# FEATURE SCALING

from sklearn.preprocessing import StandardScaler

scaler_reg = StandardScaler()

X_train_reg_scaled = scaler_reg.fit_transform(X_train_reg)
X_test_reg_scaled = scaler_reg.transform(X_test_reg)

scaler_cls = StandardScaler()

X_train_cls_scaled = scaler_cls.fit_transform(X_train_cls)
X_test_cls_scaled = scaler_cls.transform(X_test_cls)

# DISPLAY SHAPES

print("\nRegression Training Shape:", X_train_reg_scaled.shape)
print("Regression Testing Shape:", X_test_reg_scaled.shape)

print("\nClassification Training Shape:", X_train_cls_scaled.shape)
print("Classification Testing Shape:", X_test_cls_scaled.shape)
```



The preprocessing results show that the dataset contains no missing values after handling the `Income` column, indicating that the dataset is complete and suitable for machine learning analysis. This helps ensure that the models can be trained without errors caused by incomplete records.

The dataset contains 182 duplicate rows, which indicates that some customer records appear more than once in the dataset. Although duplicates exist, the dataset still remains usable for analysis and predictive modelling.

After preprocessing, the regression dataset was divided into training and testing sets with 1792 training records and 448 testing records using 21 input features. Similarly, the classification dataset was split into 1792 training records and 448 testing records using 27 features.

The difference in the number of features between regression and classification tasks occurred because spending-related variables used to create the `Total_Spending` target variable were removed from the regression feature set to prevent data leakage. This helps ensure that the regression models produce realistic and unbiased predictions.


# **PHASE 5 — REGRESSION MODELS**

```python
# LINEAR REGRESSION

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

import numpy as np
import matplotlib.pyplot as plt

lr_model = LinearRegression()

lr_model.fit(X_train_reg_scaled, y_train_reg)

y_pred_lr = lr_model.predict(X_test_reg_scaled)

mae_lr = mean_absolute_error(y_test_reg, y_pred_lr)
mse_lr = mean_squared_error(y_test_reg, y_pred_lr)
rmse_lr = np.sqrt(mse_lr)
r2_lr = r2_score(y_test_reg, y_pred_lr)

print("LINEAR REGRESSION RESULTS")
print("MAE:", mae_lr)
print("MSE:", mse_lr)
print("RMSE:", rmse_lr)
print("R2 Score:", r2_lr)

plt.figure(figsize=(8,6))

plt.scatter(y_test_reg, y_pred_lr)

plt.xlabel("Actual Values")
plt.ylabel("Predicted Values")
plt.title("Linear Regression: Actual vs Predicted")

plt.show()
```



The Linear Regression model was used to predict customer `Total_Spending` based on demographic and purchasing-related features. The model achieved an R² Score of approximately 0.756, which indicates that around 75.6% of the variation in customer spending was successfully explained by the model.

The MAE value of approximately 191.75 indicates that the model predictions differ from the actual spending values by about 192 units on average. Similarly, the RMSE value of 294.60 shows the overall prediction error magnitude and suggests that some predictions contain moderate deviations from actual spending values.

Overall, the model demonstrates reasonably good predictive performance and is able to capture important relationships between customer characteristics and spending behavior. However, the prediction errors also indicate that customer spending is influenced by additional complex factors that may not be fully captured by a simple linear model.

From a business perspective, the Linear Regression model can help businesses estimate customer spending behavior and identify factors that influence purchasing activity. These predictions can support customer segmentation, targeted marketing campaigns, and revenue forecasting strategies.

The MAE value of approximately 385.89 and the RMSE value of 561.89 indicate that the prediction errors were relatively high compared to the Linear Regression model. This suggests that the SVR model was less effective in accurately predicting customer spending patterns for this dataset.

The low R² Score and higher error values indicate that the SVR model was unable to capture the relationships between the input variables and the target variable effectively. From a business perspective, the SVR model provides limited predictive performance for estimating customer spending behavior in this project. # **PHASE 6 — CLASSIFICATION MODELS**

```python
# LOGISTIC REGRESSION

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix

import seaborn as sns
import matplotlib.pyplot as plt

log_model = LogisticRegression(max_iter=1000)

log_model.fit(X_train_cls_scaled, y_train_cls)

y_pred_log = log_model.predict(X_test_cls_scaled)

accuracy_log = accuracy_score(y_test_cls, y_pred_log)
precision_log = precision_score(y_test_cls, y_pred_log)
recall_log = recall_score(y_test_cls, y_pred_log)
f1_log = f1_score(y_test_cls, y_pred_log)

print("LOGISTIC REGRESSION RESULTS")
print("Accuracy:", accuracy_log)
print("Precision:", precision_log)
print("Recall:", recall_log)
print("F1 Score:", f1_log)

cm = confusion_matrix(y_test_cls, y_pred_log)

plt.figure(figsize=(6,5))

sns.heatmap(cm, annot=True, fmt='d')

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Logistic Regression Confusion Matrix")

plt.show()
```



The Logistic Regression model was used to predict customer campaign response behavior using demographic and purchasing-related features. The model achieved an accuracy of approximately 87.3%, indicating that the model correctly classified a large proportion of customer responses.

The precision score of approximately 0.692 indicates that when the model predicted a positive campaign response, the prediction was correct about 69.2% of the time. However, the recall score of approximately 0.269 shows that the model identified only a limited portion of actual positive responses.

The F1 Score of approximately 0.387 suggests that the balance between precision and recall was moderate. Although the model achieved high overall accuracy, the lower recall indicates difficulty in detecting all customers who actually responded positively to the campaign.

This behavior is likely influenced by class imbalance in the dataset, where non-responding customers significantly outnumber responding customers. As a result, the model performs better at identifying negative responses than positive ones.

From a business perspective, the Logistic Regression model can help businesses identify customers who are more likely to respond to marketing campaigns. However, improvements may be needed to better capture potential positive responders and improve marketing targeting effectiveness.
